You can install from PyPI (or from source locally):


```bash

```

## Load packaged datasets

The package includes the following datasets: `simdata`, `gs2020`, `simgsynth`, `turnout`, and `hh2019`.

Two ways to load:
- Unified data entry: `from fect import data; data.gs2020()`
- Module-style import: `import fect.data.gs2020 as gs2020; gs2020.load()` (or `gs2020.df` if preloaded)


````python
from fect import data as fdata
import pandas as pd

# Load a few datasets and show shapes
print('gs2020:', fdata.gs2020().shape)
print('hh2019:', fdata.hh2019().shape)
print('turnout:', fdata.turnout().shape)

# Preview
fdata.gs2020().head()
````
